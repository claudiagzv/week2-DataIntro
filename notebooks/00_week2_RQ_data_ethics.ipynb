## 1. Research Question

- **Main research question:** How effectively can machine learning models predict the antiparasitic activity of compounds against neglected tropical diseases using molecular structure and physicochemical properties from the MMV/ChEMBL dataset?
- **Why is this question interesting or important?** neglected tropical diseases affect millions of people and there is a need to identify potential new treatments. Machine learning could help identify relationships between molecular properties and antiparasitic activity, potentially allowing compounds with promising activity to be identified more efficiently.
- **Who might use or care about the answer?** Pharmaceutical and drug-discovery researchers, computational chemists, medicinal chemists and organisations working on neglected tropical diseases could be interested in whether molecular data can be used to predict antiparasitic activity.
- **Possible sub-questions:** 

    -Which molecular properties are most strongly associated with antiparasitic activity?

    -How accurately can different machine learning models predict pIC50?

    -Does model performance differ between different neglected diseases?

    -Can molecular descriptors be used to identify compounds with potentially broad antiparasitic activity?


## 2. Dataset Reference

- **Dataset name:** MMV/ChEMBL Neglected Tropical Disease IC50 Dataset
- **Source link:** (https://www.ebi.ac.uk/chembl/)
- **Date accessed:** 2026-09-11
- **Who published or collected it?** The underlying experimental data are from the ChEMBL database, developed and maintained by EMBL-EBI. The specific dataset used in this project was extracted and processed by Ant Chubb and Darren Fayne for the DCU data analytics project.
- **Why did you choose this dataset?** This dataset was chosen because it contains molecular structures, physicochemical properties and experimental bioactivity measurements, making it suitable for a machine learning project investigating relationships between chemical structure and antiparasitic activity. It also provides data for several neglected tropical diseases, allowing activity against individual diseases or multiple diseases to be investigated.
- **Evidence or notes:** The dataset contains molecular descriptors including molecular weight, logP, hydrogen-bond acceptors/donors, polar surface area, rotatable bonds and Lipinski Rule-of-Five violations. It also contains IC50 measurements for several neglected tropical diseases and HepG2 CC50 measurements for cytotoxicity/selectivity analysis.

    Ant confirmed that repeated IC50 measurements were averaged during processing and that the resulting averages were consistent with the values reported in the original CSV. ### 3. Who is represented?

**Answer:** 

*Known evidence -* The dataset represents chemical compounds and their experimentally measured activity against neglected tropical disease pathogens.
